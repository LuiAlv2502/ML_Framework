# Pruebas del paquete MVC

Este notebook ejecuta pruebas automatizadas sobre la carga genérica de CSV, EDA y los algoritmos no supervisados del proyecto. Las pruebas se encuentran en `../tests/test_mvc.py` para que también puedan ejecutarse fuera de Jupyter.

## 1. Configurar el entorno de pruebas

El notebook usa `pytest` para ejecutar la suite y `pytest-cov` para medir cobertura. Ejecute la instalación solo si el kernel no tiene estas dependencias.

```python
# Descomente si el kernel no cuenta con las dependencias:
# %pip install pytest pytest-cov

from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SRC = ROOT / "src"
TESTS = ROOT / "tests"

if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

print(f"Proyecto: {ROOT}")
print(f"Pruebas: {TESTS}")
```

## 2. Definir funciones objetivo para validar

Las pruebas validan métodos reales de `DataModel` y `DataFrame`: carga de CSV, restauración, imputación, normalización, PCA, HAC, K-Means, t-SNE y UMAP.

```python
from model import DataModel

modelo = DataModel(ROOT / "data" / "drug200.csv")
print("Dimensiones del dataset de prueba:", modelo.dimensiones())
print("Algoritmos cubiertos: PCA, HAC, K-Means, t-SNE y UMAP")
```

## 3. Escribir pruebas unitarias con pytest

La suite está en `tests/test_mvc.py`. Cada prueba usa aserciones para validar valores de retorno, dimensiones, métricas y restauración de datos.

```python
archivo_pruebas = TESTS / "test_mvc.py"
assert archivo_pruebas.exists(), f"No existe {archivo_pruebas}"
print(archivo_pruebas.read_text(encoding="utf-8"))
```

## 4. Agregar pruebas parametrizadas y casos límite

La suite usa `@pytest.mark.parametrize` para ejecutar HAC y K-Means, así como t-SNE y UMAP. También verifica datos vacíos e imputación de valores nulos.

```python
contenido = archivo_pruebas.read_text(encoding="utf-8")
assert "@pytest.mark.parametrize" in contenido
assert "test_pca_rechaza_dataframe_vacio" in contenido
assert "test_imputar_nulos_y_normalizar" in contenido
print("Pruebas parametrizadas y casos límite presentes.")
```

## 5. Ejecutar pruebas y revisar resultados

La siguiente celda ejecuta toda la suite con salida detallada. Si una prueba falla, `pytest` mostrará la traza para depuración.

```python
import subprocess

resultado = subprocess.run(
    [sys.executable, "-m", "pytest", "tests/test_mvc.py", "-q"],
    cwd=ROOT,
    capture_output=True,
    text=True,
)
print(resultado.stdout)
print(resultado.stderr)
assert resultado.returncode == 0, "Hay pruebas fallidas."
```

## 6. Medir cobertura con pytest-cov

La cobertura identifica qué líneas de `src/` se ejecutan durante las pruebas. Las líneas no cubiertas son candidatas para nuevas pruebas.

```python
cobertura = subprocess.run(
    [
        sys.executable,
        "-m",
        "pytest",
        "tests/test_mvc.py",
        "-q",
        "--cov=src",
        "--cov-report=term-missing",
    ],
    cwd=ROOT,
    capture_output=True,
    text=True,
)
print(cobertura.stdout)
print(cobertura.stderr)
assert cobertura.returncode == 0, "No se pudo generar la cobertura."
```
