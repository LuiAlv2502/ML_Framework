# MVC de análisis de datos en Jupyter

Estructura del proyecto:

```text
IA_Laboratorios/
├── data/                    datasets CSV
├── src/
│   ├── dataframe_desarrollado.py  EDA y adaptador de análisis
│   ├── model.py, view.py, controller.py
│   ├── unsupervised/        PCA, HAC, K-Means, t-SNE y UMAP
│   └── supervised/          espacio para futuros modelos supervisados
└── notebooks/               este notebook
```

Arquitectura:

```text
            JUPYTER
               |
               v
        DataController
         /              \
        v                v
    DataModel <------ DataView
        |
        v
   DataFrame (EDA compartido)
        |
        +-- unsupervised/   reducción dimensional y clustering
        +-- supervised/     clasificación y regresión (futuro)
```

- **DataModel:** carga cualquier CSV, restaura y delega las operaciones a `DataFrame`.
- **DataFrame:** concentra EDA común y conserva una API única para el MVC.
- **unsupervised:** contiene la implementación de PCA, HAC, K-Means, t-SNE y UMAP.
- **supervised:** paquete separado para agregar modelos supervisados sin mezclar responsabilidades.


````python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "src").is_dir():
    project_root = project_root.parent
if not (project_root / "src").is_dir():
    raise FileNotFoundError("No se encuentra la carpeta src del proyecto.")

src_path = str(project_root / "src")
if src_path not in sys.path:
    sys.path.insert(0, src_path)

from model import DataModel
from view import DataView
from controller import DataController
````

````python
# MODEL: se inicia vacío; el archivo se carga desde la interfaz (campo "Archivo").
model = DataModel()

# VIEW
view = DataView()
view.ruta_archivo.value = str(project_root / "data" / "drug200.csv")

# CONTROLLER
controller = DataController(model, view)
````

````python
# Iniciar aplicación MVC
controller.iniciar()
````
